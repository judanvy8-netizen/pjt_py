```python
# Human클래스에 setinfo()메서드 추가

class Human:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
        
    def who(self):
        print(f'이름 : {self.name}, 나이 : {self.age}, 성별 : {self.gender}')
        
    def setinfo(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

hong = Human('익명', 0, '모름')
hong.who()

# hong객체의 정보를 수정해야할 경우!
hong.setinfo('홍길동', 25, '남자')
hong.who()    
```

### 클래스 Class

#### 1. module vs class vs instance

1. module : 확장자가 ~.py로 끝나는 파일(클래스, 속성(필드), 메서드)
2. class : 새로운 객체(자료형)를 만드는 방법(설계도, 틀...)
3. instance : 클래스로부터 생성된 객체를 인스턴스라고 한다.

* function(함수)
* method = function이 클래스안에 존재
    - def 함수명(self,.)
    
#### 2. 클래스 내부의 메서드(함수) 선언

1. 메서드는 일반함수(def 함수명() : pass)와 동일하게 정의하지만, 다른 점은 `첫번째 인수로 self로 정의`해야 한다. 즉, `def 함수명(self, 매개값....)`의 형태로 정의해야만 한다.
2. 이 self예약어는 java에서 this처럼 `객체 자기자산을 가리키는 것`으로 인스턴스(객체) 자신을 가리킨다.
3. 정확하게는 `객체의 메모리 주소`를 참조한다.
4. 개개의 인스턴스는 self를 이용해서 자신에게 접근할 수 있다.


#### 3. 생성자(`__init__()`, constructor)
1. 파이썬에서는 객체(인스턴스)가 생성될 때 자동으로 실행되는 특수한 메서드를 생성자메서드라고 한다.
2. 생성자메서는 객체가 생성될 때 맨 처음에 자동으로 실행되는 메서드로이다.
3. 파이썬에서는 생성메서드는 오직 `1개만 정의`할 수 있다.(여러개 생성하면 맨 마지막 생성자만 유효)

#### 4. 소멸자(`__del__()`)
1. 파이썬에서는 메모리나 기타자원을 자동으로 관리해 주는데 객체가 소멸될 때 자동으로 실행되는 메서드이다.
2. 만약 객체가 소멸될 때 수행될 로직이 있을 경우에 소멸메서드에 해당 로직으로 구현해 주면 된다. 없을 경우에는 특별히 정의하지 않아도 된다.

#### 5. 상속(inheritance)
1. 파이썬에서는 `다중상속이 가능`하다.
2. 상속의 장점은 코드의 재사용이다.
3. 자식클래스는 상속을 해 준 부모클래스의 `모든 기능을 사용`할 수 있다.
4. 자식 클래스에서 필요한 기능을 추가하거나 부모의 기능을 재정의해서 사용할 수 있다.


#### 6. 오버라이드(override)
* 자식(Child, sub)클래스에서 부모(Parent, super)클래스에 정의된 메서드를 자식클래스에서 재정의하는것을 오버라이드라고 한다.

#### 7. 다형성(Polyorphism)

1. 상속관계내의 다른 클래스들의 인스턴스들이 동일 멤버함수를 호출하는 것에 대해 다른 형태로 실행되도록 정의하는 기능을 말한다.
2. 다형성의 장점은 적은 코딩으로 다양한 인스턴스들에게 유사한 작업이 가능하도록 한다.
3. 코드의 가독성을 높인다.

#### 4.Static

1. 일반적으로 객체의 멤버(변수, 메서드)에 접근하려면 객체(인스턴스)가 생성된 후에 접근할 수 있다.
2. 하지만 객체(인스턴스)를 생성하지 않고 클래스로 직접접근해서 호출할 수 있도록 `static method`와 `class method`가 있다.
3. 파이썬에서는 기본적으로 모든 변수와 메서드에 접근할 수 있다. 하지만 `캡슐화(은닉화)`를 통해서 특정변수와 메서드에 접근할 수 있도록 제한할 수가 있다.
4. java처럼 완벽한 캡슐화(은닉화)를 제공되지 않는다.

##### 1) static method

* 정적메서드는 메서드정의문자 바로 앞에 `@static method`라는 메타정보를 지정한다.
* 이 때 주의할 점은 매개변수에 `self를 정의하지 않는다.`
> `@staticmethod`<br>
> `def 메서드(arg1, arg2, .... ) : ` pass

##### 2) class method
* 클래스메서드 특정객체에 대한 작업처리를 하는 것이 아니라 `클래스 전체에 공통으로 사용되는 메서드`로서 `첫번째 매값에 cls`로 전달되는 메서드를 말한다.
* 즉, `def 메서드명(cls, arg1, arg2....):`의 형태로 정의되는 메서드이다.
* 클래스 메서드는 인스턴스(객체)와 무관하게 `클래스 namespace에 존재하는 메서드`이다.
* 클래스 메서드에 접근하기 위해서는 `클래스.클래스 메서드()`형태로 접근한다.
* 클래스메서드는 정적 메서드와 비슷하지만 약간의 차이점이 있다. 클래서 메서드는 메서드 선언문 바로 위에 `@classmethod`라는 메타정보를 선언한다.
> `@classmethod`<br>
> `def 메서드(cls, arg1, arg2, .... ) : ` pass

##### 3) staticmethod와 classmethod의 차이점
* 클래스메서드와 정적메서드는 `클래스로 직접접근도 가능`하고 `객체에서 접근도 가능`하다.
* 차이점은 상속에서 차이가 난다. 정적메서드의 경우 부모클래스 속성을 가져오지만 class메서드는 현재 클래서 속성으로 가져온다.

#### 4) `@데코레이터`
파이썬은 데코레이라는 기능을 제공한다. 메서드를 장식한다는 의미

```python
class 클래스():
    def xxx(self):
        print('객체메서드가 호출되었습니다.')
        
    @staticmethod
    def yyy():
        print('정적메서드가 호출되었습니다.')
        
    @classmethod
    def zzz(cls):
        print('클래스메서드가 호출되었습니다.')
        
```

```python
# 1. 객체생성없이 직접 클래스로 메서드에 접근하기

# 클래스.xxx() 에러

클래스.yyy()
클래스.zzz()
```

```python
# 2. 객체를 생성한 후에 접근 : 
a = 클래스()

a.xxx()

# java와 다르게 yyy(), zzz()는 객체로 접근할 수가 있음
a.yyy()
a.zzz()
```

```python
# 3. stacit method와 variable

# 1) staica 메서드 : self가 없을 경우
# 주의할 점 : 동일이름의 메서드가 있을 경우에는 매개변수와 상관없이 마지막에 정의된 메서드로 호출


class Calc:
    
    @staticmethod
    def add(a,b):
        return f'정적메서드 더하기 = {a+b}'
    
    
    
    def mul(self, a,b):
        return f'일반메서드 곱하기 = {a*b}'
    
    @staticmethod
    def mul(a,b):
        return f'정적메서드 곱하기 = {a*b}'
    
a = Calc()

a.add(1,2)

a.mul(1,2)
```

#### 6. 추상클래스

* 파이썬에서도 java처럼 `추상클래스(abstract class)`를 제공한다.
* 추상 클래스는 메서드의 목록만 가진 즉, 실행문이 없는 메서드로 이르만 가진 클래스이다.
* 상속받은 클래스에서 `메서드의 구현을 강제화`하기 위해서 사용한다.
* 추상클래스를 만들려면 `import abc`처럼 `abc(Abstract Base Class)`모듈을 가져와야 한다.
* 그리고 클래스의 소괄호()안에 `metaclass=ABCMeta`를 지정하고 메서드선언문위에 `@abstractmethod`를 붙여서 추상메서드로 지정한다.
> `from abc import *` <br>
> class 추상클래스명(`metaclass=ABCMeta`): <br>
>       @abstractmethod  <br>
>       def 추상메서드() : pass

```python
# 1. 추상클래스 만들기

from abc import *

class StudentAbstract(metaclass = ABCMeta):
    
    @abstractmethod
    def study(self):pass
    @abstractmethod
    def goto_school(self):pass
    
    def xxx(self):
        print('운동을 합니다.')
        

class Student(StudentAbstract):
    def study(self):
        print('공부를 합니다.')  # 자식 클래스에서 강제로 재정의하기 위해 
    def goto_school(self):
        print('학교에 갑니다.')
        
a = Student()
# b = StudentAbstract()
a.study()
a.goto_school()
a.xxx()

# 추상메서드를 빈 메서드로 두는 이유
# 추상클래스는 객체(인스턴스)를 만들수 없다.
# 따라서, 추상클래스는 인스턴스로 만들 때 사용하지 않고, 오직 상속에서만 사용된다.
# 구현 클래스 즉, 자식클래스에서는 추상메서드로 정의된 메서드는 반드시 구현해야 한다.
# 추상메서드가 구현되지 않을 경우 자식 객체를 생성할 때 에러가 발생한다.
```

#### 8. 캡슐화(은닉화)

##### 비공개(공유) 속성 사용하기

1. 클래스 속성도 인스턴스속성처럼 비공유속성(비공개)으로 만들 수가 있습니다.
2. 클래스 변수를 만들 때 변수의 이름 앞에 `언더바 2개로 시작` 즉 `__변수명, __메서드`으로 시작
3. 따라서, 비공개 속성은 `클래스안에서만 접근할 수 있고 바깥에서는 접근할 수 없다.`
4. 파이썬에서는 java와 같이 완벽한 비공개 즉 캡슐화하기 어렵다.
5. `__속성__`처럼 언더바 2개가 시작하고 끝나는 속성은 `캡슐화와는 상관이 없다`
6. 즉 `__속성은 비공개속성`이고, `__속성__은 공개속성`이다.

```python
class Korean:
    country = '대한민국'
    __country = 'Korea'   # java에서 private로 선언, 외부에서 getter()로 접근
    __country__ = '한국'
    
    def getCountry(self):
        return Korean.__country
    
    
kor = Korean()
print(kor.country)
print(kor.__country__)
# print(kor.__country) 오류 : 비공개 속성이니깐
kor.getCountry()
```

```python
kor.__country = "값변경하기"  # kor객체 속성이 하나 생성!
print(kor.__country)
print(kor.getCountry())
```

```python
# 캡슐화(은닉화, 비공개)된 변수는 getter와 setter를 통해서 접근이 가능하다.

class Korean:
    country = '대한민국'
    __country = 'Korea'   # java에서 private로 선언, 외부에서 getter()로 접근
    __country__ = '한국'
    
    
    def getCountry(self):
        return Korean.__country
    
    def setCountry(self, country):
        Korean.__country = country
        
kim = Korean()

print(kim.getCountry())
kim.setCountry('Republic of Korea')
print(kim.getCountry())
```

```python
# @staticmethod와 @classmethod 차이

class Parent:
    word = 'current date :'
    
    @staticmethod
    def now_static():
        return(Parent())
    
    @classmethod
    def now_class(cls):
        return(cls())
    
class Child(Parent):
    word = '현재 날짜 :'
    
child = Child()

print(child.now_static().word)
print(child.now_class().word)
```
