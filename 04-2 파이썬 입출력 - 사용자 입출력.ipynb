따옴표로 둘러싸인 문자열은 + 연산과 동일하다

```python
print("life" "is" "too short")  # 1번 + 연산
```

```python
print("life"+"is"+"too short")  # 2번
```

문자열 띄어쓰기는 쉼표로 한다

```python
print("life", "is", "too short")
```

쉼표(,)를 사용하면 문자열 사이에 공백이 자동으로 추가된다.

sep 매개변수로 구분자 설정하기

```python
print("2026", "09", "07", sep="-")
```

```python
print('안녕하세요', '파이썬', sep='▧')
print('안녕하세요', '파이썬', end='')  # end의 default 값은 '\n' --> 줄바꿈 --> 공백
print('어려워요') # 붙어서 출력이 되는 이유는?
```
