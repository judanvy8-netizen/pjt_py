### 파일 생성하기

모드   | 설명 |
-------|-------------------------------------------|
| `r` | 읽기 모드 , 파일을 읽기만 허용 |
| `w` | 쓰기모드 , 파일에 내용을 쓸 때 사용|
| `a` | 추가모드 - 파일의 마지막에 새로운 내용을 추가 시킬 때 사용 |

```python
# newfile.py
f = open("새파일.txt", 'w')  # 쓰기 모드
f.close()
```

### 파일을 읽는 여러 가지 방법

* readline 함수 이용하기
* readlines 함수 사용하기
* read 함수 사용하기
* 파일 객체를 for 문과 함께 사용하기

readline 함수 이용하기

* 파일에서 한줄 읽기

```python
# readline_test.py
f = open("C:/doit/새파일.txt", 'r')
line = f.readline()
print(line)
f.close()
```

위 프로그램을 실행했을 때 새파일.txt 파일의 가장 첫 번째 줄이 화면에 출력될 것이다.

만약 모든 줄을 읽어 화면에 출력하고 싶다면 다음과 같이 작성하면 된다.

```python
# readline_all.py
f = open("C:/doit/새파일.txt", 'r')
while True:
    line = f.readline()
    if not line:
        break
    print(line)
f.close()
```

readlines 함수 사용하기

```python
# readlines.py
f = open("C:/doit/새파일.txt", 'r')
lines = f.readlines()
for line in lines:
    print(line)
f.close()
```

줄 바꿈(\n) 문자 제거하기

```python
f = open("C:/doit/새파일.txt", 'r')
lines = f.readlines()
for line in lines:
    line = line.strip()  # 줄 끝의 줄 바꿈 문자를 제거한다.
    print(line)
f.close()
```

read 함수 사용하기

```python
# read.py
f = open("C:/doit/새파일.txt", 'r')
data = f.read()
print(data)
f.close()
```

파일 객체를 for 문과 함께 사용하기

```python
# read_for.py
f = open("C:/doit/새파일.txt", 'r')
for line in f:
    print(line)
f.close()
```

### 파일에 새로운 내용 추가하기

* 쓰기 모드('w')로 파일을 열 때 이미 존재하는 파일을 열면 그 파일의 내용이 모두 사라진다.
* 하지만 원래 있던 값을 유지하면서 새로운 값만 추가해야 할 경우도 있다.
* 이런 경우에는 파일을 추가 모드('a')로 열면 된다.

```python
# add_data.py
f = open("C:/doit/새파일.txt",'a')
for i in range(11, 21):
    data = f"{i}번째 줄입니다.\n"
    f.write(data)
f.close()
```

with 문과 함께 사용하기

* 파일을 열면(open) 항상 닫아(close) 주어야 한다
* 파이썬의 with 문이 바로 이런 역할을 해 준다

```python
# file_with.py
with open("foo.txt", "w") as f:
    f.write("Life is too short, you need python")
```

```python
with open("foo.txt", "w") as f:
    f.write("Hello")
    print(f.closed)  # False (아직 열려 있음)

print(f.closed)  # True (with 블록을 벗어나 자동으로 닫힘)
```

with 문 안에서 만든 변수 사용하기

* with 문 안에서 만든 변수는 with 블록이 끝난 후에도 사용할 수 있다. 다음 예를 보자.

```python
with open("test.txt", "w") as f:
    content = "Hello, Python!"
    f.write(content)

print(content)  # "Hello, Python!" 출력
```

파일 처리 시 주의사항

* 한글이 포함된 파일을 다룰 때는 인코딩을 명시하는 것이 좋다.

```python
# 한글 파일 쓰기
with open("한글파일.txt", "w", encoding="utf-8") as f:
    f.write("안녕하세요, 파이썬!")

# 한글 파일 읽기
with open("한글파일.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)
```
