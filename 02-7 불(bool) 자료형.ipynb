### 불(Bool, Boolean) 자료형

* bool자료형이란 `참과 거짓을 나타내는 자료형`이다.
* bool자료형은 2가지 값인 `True` or `False`값만 가질 수 있다.
* 주의할 점은 True와 False는 파이썬의 예약어로서 첫글자는 반드시 대문자이어야 한다.

### 불 자료형은 어떻게 사용할까?

```python
a = True
b = False
```

```python
type(a)
```

```python
type(b)
```

불 자료형은 조건문의 반환값으로도 사용된다.

```python
1 == 1
```

```python
2 > 1
```

```python
2 < 1
```

자료형의 참과 거짓

```python
a = [1, 2, 3, 4]
while a:
    print(a.pop()) # 리스트에서 마지막 요소를 제거
    print(a)
```

```python
if [1, 2, 3]:
    print("참")
else:
    print("거짓")
```

위에 코드를 해석하면

만약 [1, 2, 3]이 참이면 "참"이라는 문자열을 출력하고, 그렇지 않으면 "거짓"이라는 문자열을 출력하라.

불 연산

* bool 함수를 사용하면 자료형의 참과 거짓을 정확하게 확인할 수 있다.

```python
bool('python')
```

```python
bool('')
```

```python
bool([1, 2, 3])
```

```python
bool([])
```

```python
bool(0)
```

```python
bool(3)
```

```python
bool(1)
```

논리 연산자

* and
* or
* not

and 연산자
* and 연산자는 양쪽 조건이 모두 참일 때만 True를 반환한다.

```python
True and True
```

```python
True and False
```

```python
False and True
```

```python
False and False
```

or 연산자
* or 연산자는 양쪽 조건 중 하나라도 참이면 True를 반환한다.

```python
True or True
```

```python
True or False
```

```python
False or True
```

```python
False or False
```

not 연산자
* not 연산자는 참과 거짓을 반대로 바꾼다.

```python
not True
```

```python
not False
```

```python
not 1
```

```python
not 0
```

```python
x = 5
y = 10
x > 0 and y > 0
```

```python
x > 10 or y > 5
```
