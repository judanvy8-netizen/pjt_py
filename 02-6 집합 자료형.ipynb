### 집합자료형(set)

* 집합자료형은 파이썬 v2.3부터 지원하기 시작한 자료형이다.
* 이 자료형은 집합과 관련된 작업을 보다 쉽게 처리할 수 있도록 만들어진 자료형인다.
* 리스트나 튜플은 순서가 있기 때문에 인덱스를 통해서 값을 가져올 수 있지만, set자료형은 `순서와 key가 없기` 때문에 인덱스나 key로 값을 가져올 수 없다.
* 따라서, set자료형을 값을 가져오기 위해서는 자료형을 `list or tuple 자료형으로 변환 후에 인덱스로 값을 가져올 수 있다.`
* set자료형은 `집합관련 연산`이나 `중복을 제거하기 위한 필터`역할로 종종 사용한다.
* set자료형으로 선언하려면 `중괄호안에 값을 가진 객체로 선언`하거나 `set()이라는 생성자`로 선언한다.
* 생성자 set()에 iterable한 객체를 전달하면 자동으로 set자료형으로 변환해 준다.

#### 집합자료형의 특징

1. `중복을 허용하지 않고 집합안에서  unique한 값`을 가진다.
2. `순서가 없다`
3. set자료형의 선언은 `중괄호{} or set()`로 선언한다
4. 중괄호로 선언하는 것은 dict와 유사하지만 set자료형은 key가 없고, `값만 존재`한다.
5. set자료형의 내부요소 즉 값은 다양한 값을 함께 가질 수 있지만 `mutable한 자료는 가질 수 없다.`

```python
print(type({}), type({1}))

# 1. set자료형 선언하기(1) - 중괄호 내우베 값을 가진 객체로 선언
s = {1,2,3,4}
print(s, type(s))

# 2. set자료형 선언하기(2) - 생성자메서드 set()으로 선언
s = set()
print(s, type(s))

# 3. set자료형 선언하기(3) - 생성자메서드 set으로 선언(반복자)
s1 = set([1,2,3,4])
s2 = set((1,2,3,4))
s3 = set('1234')
print(type(s1), type(s2), type(s3))
s4 = set({1:'a', 2:'b'})
print(s4, type(s4))
# key만 set타입으로 출력

# 4. 중복된 값이 있을 경우 : 중복 제거 + 순서x
s = {1,5,1,1,"1",1,2,3,4,5}

print(s)
```

```python
s3 = {1, 2, 3}
s3
```

```python
s4 = {'a', 'b', 'c'}
s4
```

```python
s = {}
print(type(s))
```

### 교집합, 합집합, 차집합 구하기

```python
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
```

교집합 구하기

* '&'  사용  또는 ntersection 함수를 사용

```python
s1 & s2
```

```python
s1.intersection(s2)
```

합집합 구하기

* '|'  사용 또는union 함수를 사용

```python
s1 | s2
```

```python
s1.union(s2)
```

차집합 구하기

* '-' 사용 또는 difference 함수를 사용
* 왼쪽에 있는 집합에서 오른쪽 집합은 뺀다

```python
s1 - s2
```

```python
s2 - s1
```

```python
s1.difference(s2)
```

```python
s2.difference(s1)
```

집합 자료형 관련 함수

* add
* update
* remove
* discard

값 1개 추가하기 - add

```python
s1 = set([1, 2, 3])
s1.add(4)
s1
```

값 여러 개 추가하기 - update

```python
s1 = set([1, 2, 3])
s1.update([4, 5, 6])
s1
```

특정 값 제거하기 - remove

```python
s1 = set([1, 2, 3])
s1.remove(2)
s1
```

특정 값 제거하기 - discard

emove와 비슷하지만, 존재하지 않는 값을 제거하려 해도 오류가 발생하지 않는다.

```python
s1 = set([1, 2, 3])
s1.discard(2)
s1
```

```python
s1.discard(4)  # 없는 값이어도 오류가 발생하지 않음
s1
```

모든 값 제거하기 - clear

```python
s1 = set([1, 2, 3])
s1.clear()
s1
```

## 집합 자료형은 어디에서 사용하나요?

* ### 데이터 분석 : wordcount  --> 빅 데이타 분석

```python
# wordcount 데이터 분석

l = ['read', 'a', 'book', 'write', 'a', 'book']
print(l)

s = set(l)
print(s, len(s))
```

```python
# 중복을 제외한 단어 개수

s = 'read a book write a book or read hangul write hangul'
print(s)

l = s.split()
print(l)
s = set(l)
print(s, len(s))
```
