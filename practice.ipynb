### A. Jupyter Notebook  기본 사용법

#### 1. 새 파일 생성

1. 프로젝트 폴더 선택 --> [+] --> 'jupyter notebook' 클릭 --> new jupyter notebook --> 이름 입력 --> 새파일 생성.
2. 생성된 파일의 이름을 변경하려면 제목 선택 --> rename --> 이름을 변경하면 된다.

#### 2. 셀의 타입
셀의 타입 `Code 타입, Markdown 타입`이 있다. `Code타입은 파이썬의 코드를 실행`할 수 있는 셀이다. 기본적인 생성 셀은 Code타입으로 생성된다. Markdown타입은 `주석으로 셀의 내용`을 작성 할 수 있다.

#### 3. 셀 실행
1. `shift + enter`(현재 셀 실행 후 다음셀로 이동), `ctrl+enter`(현재 셀 실행만)
2. 실행한 후에 셀 아래쪽에 실행결과가 출력이 된다.
