### 변수

* 변수란 ? `자료형의 값을 저장하는 메모리 공간`을 말한다.
* 파이썬에서 변수를 만드는 방법은 `변수명 = 값` 즉, `대입연산자(assignment), =`을 사용한다.
* C, Java처럼 변수의 자료형과 함께 사용하지 않는다.
* 파이썬에서는 변수에 저장된 값을 가지고 스스로 판단해서 자료형을 알아낸다.
* 파이썬에 사용하는 `변수는 객체(object)의 메모리주소를 가지고 있다.`즉, `객체가 저장되어 있는 메모리 주소`를 가리키는 것이다.
* 파이썬에서는 숫자(정수) 3은 일반적으로 상수값이지만, 파이썬에는 `3은 정수형 객체(object)`이다.
* 예를 들어서 `print(type(3))`의 결과는 `<class int>`이다. 즉, `3은 int클래스의 객체`임을 알 수 있다.

리스트를 복사할 때

* 얕은 복사 : b = a
* 깊은 복사 : 슬라이싱 이용 또는 copy() 이용

얕은 복사

```python
a = [1, 2, 3]
b = a
```

깊은 복사 --> 1. [:] 이용하기

```python
a = [1, 2, 3]
b = a[:]
a[1] = 4
a
```

a 리스트 값을 바꾸더라도 b 리스트에는 영향이 없다.

```python
b
```

깊은 복사 --> 2. copy 모듈 이용하기

```python
from copy import copy # copy() --> 외장함수
a = [1, 2, 3]
b = copy(a)
```

False를 반환하므로 b와 a가 가리키는 객체는 서로 다르다.

```python
b is a
```

### copy 함수 사용하기 --> list 내장함수인 copy()

```python
a = [1, 2, 3]
b = a.copy()
b is a
```

1. import 모듈명 사용할때 --> 함수 사용방법

```python
import copy   # 모듈 copy 를 import
a = [1,2,3]
b = copy.copy(a)
print(a)
print(b)
```

2. from 모듈명 import 함수명 사용시 --> 함수 사용 방법

```python
from copy import copy   # 모듈 copy 에서 copy 함수 import
a = [1,2,3]
b = copy(a)
print(a)
print(b)
```

### 변수를 만드는 여러 가지 방법

튜플로 a, b에 값을 대입할 수 있다.

```python
a, b = ('python', 'life')
print(a)
print(b)
```

튜플은 괄호를 생략해도 된다.

```python
(a, b) = 'python', 'life'
print(a)
print(b)
```

여러 개의 변수에 같은 값을 대입할 수도 있다.

```python
a = b = 'python'
print(a)
print(b)
```

파이썬에서는 이 방법을 사용하여 두 변수의 값을 간단하게 바꿀 수 있다.

```python
a = 3
b = 5
a, b = b, a
```

```python
b
```

## 변수를 복사하는 방법

```python
# 얕은복사 (메모리 주소가 같다)
l1 = [1,2,3]
l2 = l1

# 깊은복사 (메모리 주소가 다르다)
l3 = l1[:]

# 깊은복사
l4 = l1.copy()

print(l1, l2, l3, l4)
print(id(l1), id(l2), id(l3), id(l4))
```

### 증감, 할당 연산자

* = : 왼쪽변수에 오른값을 할당
* `+=` : 왼쪽변수에 오른쪽값을 더한 후, 그 결과를 왼쪽변수에 할당
* `-=` : 왼쪽변수에 오른쪽값을 뺀 후, 그 결과를 왼쪽변수에 할당
* *= : 왼쪽변수에 오른쪽값을 곱한 후, 그 결과를 왼쪽변수에 할당
* /= : 왼쪽변수에 오른쪽값을 나눈 후, 그 결과를 왼쪽변수에 할당
* %= : 왼쪽변수에 오른쪽값을 나눈 후, 그 나머지 결과를 왼쪽변수에 할당

```python
a = 1
print(a)
# a++    # 자바 ++ -- --> 파이썬은 에러
a += a
a
```
