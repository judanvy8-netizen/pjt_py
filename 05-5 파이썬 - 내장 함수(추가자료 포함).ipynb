```python
# 9. filter(func, iterable)
# filter?
# 첫 번째 인수로 함수, 두번째 인수로 iterable자료형을 전달 받음
# 그 중에서 조건에 맞는 값만 리턴하는 함수

# list자료형에서 숫자일 경우 양수값만 리턴하는 함수

def positive(nums):
    result = []
    for i in nums:
        if i>0:result.append(i)
    return result
print(positive([-2,-1, 0,1,2,3]))

# filter 함수

def positive(nums):
    return nums>0

print(list(filter(positive,[-2,-1, 0,1,2,3])))

# lambda함수

print(list(filter(lambda x:x>0, [-2,-1, 0,1,2,3])))
```

```python
# 10. map(func, iterable)
# 함수와 iterable을 전달 받아서 각 요소가 함수에 수행된 결과를 리턴

# 반복자료형의 각 요소에 2배를 리턴함수

# 1) 사용자 함수
def two_times(nums):
    result = []
    for i in nums:
        result.append(i*2)
    return result

print(two_times([-2, -1,0,1,2,3]))

# map함수
def two_times(num):
    return num*2

print(list(map(two_times, [-2, -1, 0, 1, 2, 3])))

# lambda함수

print(list(map(lambda x:x*2, [-2,-1,0,1,2,3])))
```

```python
# lambda함수와 range

print(list(map(lambda x:x*3, range(10))))
```

```python
# filter vs map

print(list(filter(lambda x:x>0, [-2,-1,0,1,2])))
print(list(map(lambda x:x>0, [-2,-1,0,1,2])))
print(list(map(lambda x:x*3, [-2,-1,0,1,2])))
```

```python
# 14. isinstance(객체, 클래스)
# 첫번째 인수로 객체, 두번째 인수로 클래스 전달 받아서
# 객체가 2번째 인수의 클래스로 생성된 객체인지를 리턴(True, False)

class Person: pass
class Korean: pass
class Calculator: pass

p = Person()
k = Korean()
c = Calculator()

print(isinstance(p, Person))
print(isinstance(p, Korean))
```

```python
# 15. issubclass(cls, tuple(class...)) : 자식클래스인지 여부

class Men(Korean):pass

print(issubclass(Men, (Korean))) # True
print(issubclass(Men, (Calculator))) # False
```

```python
# 22. zip(iterable) : 동일한 갯수의자료형을 묶어서 tuple로 리턴

# 1) 동수일 경우
a = [1,2,3]
b = [4,5,6]
print(list(zip(a,b)))

# 2) 동수가 아닐 경우
print(list(zip([1,2],[1,2,3])))

# 3) 자료형이 다를 경우 : 가능
print(list(zip((1,2,3),[1,2,3])))
print(list(zip(('abcdefg'),[1,2,3])))

# 4) 2가 이상의 반복자료형
print(list(zip([1,2,3],'abcdefg',range(5), range(10,30,6))))
```
