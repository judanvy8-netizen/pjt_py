### 모듈 Module

#### 1. 모듈의 정의

* 레벨 : 패키지(디렉토리/폴더) > 모듈(~.py파일) > 클래스(변수, 메서드) > 함수(Function)
* 함수 : 파일내에서 일정한 작업을 수행하는 코드 블럭을 말한다.
* 모듈 : 함수나 변수 or 클래스들을 모아 놓은 파일(`확장자가 py`)이다.
    - 유사하거나 연관된 작업을 하는 함수나 변수들을 모아서 하나의 파일에 저장한 후에 재사용하기 위해 정의한다.
    - 파일의 이름에서 확장자(py)를 제외한 것이 모듈의 이름이다.
* 패키지 : 여러개의 모듈을 모아 놓은 디렉토리이다.

#### 2. 모듈을 사용하는 목적

* `코드의 재사용` : 모듈이 없다면 자주 사용하는 함수를 매번 작성해야 하기 때문에 개발 및 유지보수가 어렵게 된다.
* 프로그램개발시에 전체 코드를 여러 모듈단위로 분리하여 설계함으로써 `작업의 효율성을 향샹`시킬 수 있다.
* 별도의 이름공간(Namespace)을 제공함으로써 동일 이름의 변수나 함수들이 각각의모듈마다 `독립적으로 사용`할 수 있다.

#### 3. 모듈의 종류

1. `표준모듈` : 파이썬에서 기본적으로 제공하는 모듈을 말한다. `math, string, sys....`등이 있다.
2. `사용자정의모듈` : 개발자가 직접 작성하여 만든 모듈을 말한다.
3. `써드파티모듈` : 다른업체나 개인이 만들어서 제공하는 모듈을 말한다. `numpy, pandas, matplotlib, scipy....` 등이 있다.

#### 4. 모듈의 선언

##### 1) 모듈 전체 참조
* `import 모듈명`
* `import 모듈명 as 별칭` : 해당모듈의 이름을 별칭으로 사용하거나 다른 모듈과 이름이 중복될 때, 모듈명이 너무 길때 등등 사용한다.
* `from 패키지(or 모듈명) import 함수(or 변수)` :  모듈내에서 필요한 기능만 사용할 경우 정의하는 데 이렇게 정의할 경우 코드안에서 모듈이름을 사용하지 않고 바로 함수나 변수에 접근할 수 있다.
* `from 패키지(or 모듈명) import *` : 해당 모듈에 존재하는 모든 함수나 변수를 현재 Namespace안으로 불러온다. 단, **언더바2개(언드스코어, __)로 시작하는 함수나 변수를 제외한 모든 변수나 함수**만 불러온다.
* `from 패키지(or 모듈명) import 함수(or 변수) as 별칭` : 해당 모듈내에서 정의된 이름을 별칭으로 사용

##### 2) 모듈내에서 함수를 호출하는 방법
* `모듈명.함수(매개값1, 매개값2, .... 매개값n)`

##### 3) 모듈을 참조할 수 있는 디렉토리 확인
* 기본적으로 현재 디렉토리에 있는 파일이나 파이썬 라이브러리가 저장되어 있는 디렉토리에 있는 모듈만 불러올 수가 있다.
* 현재 디렉토리를 확인하려면 `sys모듈을 import`한 후에 `sys.path`함수를 확인할 수 있다.
* 디렉토리를 현재 namespace에 추가하거나, 삭제할 경우에는 `sys.path.append()` or `sys.path.remove()`함수를 사용한다.

#### `__name__`과 `__ main__` : 모듈과 시작점에 관 얘기

* `if __name__ = __main__` : 으로 시작하는 부분을 자주 만나게 됨.
* 파이썬의 `__name__`변수는 파이썬에서 `내부적으로 사용하는 특별한 변수`이다.
* 만약에 "python hello.py"처럼 즉, 모듈을 직접실행할 경우(프롬프트에서)에는 `hello.__name__`변수에는 `__main__`이라는 문자열이 자동으로 저장된다.
* 하지만 파이썬셀이나 주피터노트북과 같이 다른 파이썬 모듈에서 `import할 경우에는 __name__변수에 자신의 이름이 저장`된다.
* 즉, import hello로 불러올 경우에는 `__name__변수에는 hello라는 문자열이 저장`된다.

#### 모듈과 시작점 알아보기

* 파이썬은 최초로 시작하는 스크립트파일과 모듈의 차이가 없음
* 스크립트 파일은 시작점도 될 수 있고, 모듈도 될 수 있음.
* name변수를 통해서 현재 스크립트파일이 시작점일지 모듈인지 판단함
* if __name__ == 'main' : 현재 스크립트파일이 프로그램의 시작점인지 판단하는 작업임
* 즉, 스크립트 파일이 메인 프로그램으로 사용될 때와 모듈롷 사용될 때를 구분하기 위한 용도임
