### 파이썬 표준 라이브러리 이란?

* 정의 : 전 세계의 파이썬 고수들이 만든 유용한 프로그램을 모아 놓은 것
* 파이썬 표준 라이브러리는 파이썬을 설치할 때 자동으로 컴퓨터에 설치된다.
* sys, re 모듈은 파이썬의 중요한 표준 라이브러리이다. sys 모듈은 04-4, re 모듈은 08장 전체에 걸쳐 다루고 있으므로 이곳에서는 설명하지 않는다.

> re (regular expressions) --> 정규 표현식 , 문자열의 패턴을 표현하는 특별한 문법 --> 암호 처리에서 사용 --> 학문 분야 --> 방대한 영역

```python
# 4. glob : 특정 디렉토리 목록을 조회(많이 사용)
import glob
print(dir(glob))
# glob.glob?
glob.glob('./data/*.*')
glob.glob('./data/*.bin')

# glob와 os.path모듈 이용하기
files = glob.glob('*')
files
type(files)  # 타입이  list 이므로

for file in files:
    if os.path.isdir(file):
        print(f'<DIR>\t\t{file}')
    else:
        print(file)
```

```python
# 5. time 모듈
import time
# 초단위 출력
time.time()
time.localtime(time.time())
tm = time.localtime(time.time())
print(tm.tm_year)
print(tm.tm_mon)
print(tm.tm_mday)
```

```python
class Person:
    name = str()
    age = int()
    job = str()

    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job

kim = Person('김씨', 42, '개발자')
park = Person('박씨', 42, '선생')
lee = Person('이씨', 42, '배우')

print(kim.name, kim.age, kim.job)
print(type(kim))
```

```python
from collections import namedtuple
# print(dir(namedtuple))

n_person = namedtuple('person', ['name', 'age', 'job'])

kim = n_person(name='김씨', age=20, job='개발자')
lee = n_person(name='이씨', age=30, job='배우')
park = n_person(name='박씨', age=20, job='선생')

print(type(kim))
print(f'{kim.name}, {kim.age}, {kim.job}') # 튜플이지만 속성형태로 접근
print()

for p in [kim, lee, park]:
    print(f'{p.name}, {p.age}, {p.job}')
```
