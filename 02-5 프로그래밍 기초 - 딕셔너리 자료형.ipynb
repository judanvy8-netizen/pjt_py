## 딕셔너리(Dictionary) 자료형

<img src = "./images/04.파이썬자료형_딕셔너리_01.png" with = "500" height ="400">

* 파이썬에서의 딕셔너리자료형이란 `key와 value를 한쌍`으로 갖는 자료형을 말한다.
* Java에서 Map과 같은 자료형
* 파이썬에서는 이러한 자료형을 `Dictionary`라고 하는데 딕셔너리는 튜플이나 리스트처럼 순차적(sequential)으로 즉, index로 해당 요소값을 처리하지 않고, `key를 통해서 value에 접근`한다

#### 딕셔너리의 특징

1. 딕셔너리의 선언은 `중괄호{} or dict()`로 한다.
>{key1:value1, key2:value2......}
2. 딕셔너리는 `immutable한 key`와 `mutable한 value`로 매핑되어 있는 순서가 없는 자료형이다.
3. 이런 이유로 key값으로는 immutable한 값을 사용할 수 있지만, mutable한 객체는 사용할 수 없다.
* 튜플은 key로 사용가능, 리스는 불가능
4. value는 중복이 될 수 있지만, key는 중복이 될 수 없다. key가 중복이 된다면 마지막인 key와 value로 결정된다.
5. 순서가 없기 때문에 인덱스로 접근할 수 없고, `key로 value에 접근`할 수 있다.
6. value는 mutable한 객체이기 때문에 `key로 접근해서 value를 수정할 수 있다.`

딕셔너리 만들 때 주의할 사항

딕셔너리에서 Key는 고유한 값이므로 중복되는 Key 값을 설정하면 하나를 제외한 나머지 것들이 모두 무시된다.

```python
a = {1:'a', 1:'b'}
a[1]     # 1: 'a' 쌍이 무시된다
```

```python
a[2]
```

또 한 가지 주의할 점은 Key에 리스트는 쓸 수 없다는 것이다.

```python
a = {[1,2] : 'hi'}
```

### 딕셔너리 관련 함수

Key 리스트 만들기 - keys

```python
a = {'name': 'pey', 'phone': '010-9999-1234', 'birth': '1118'}
a.keys()
```

dict_keys 객체를 리스트로 변환하려면 다음과 같이 하면 된다.

```python
list(a.keys())
```

dict_keys 객체는 다음과 같이 사용할 수 있다. 리스트를 사용하는 것과 별 차이는 없지만,
리스트 고유의 append, insert, pop, remove, sort 함수는 수행할 수 없다.

```python
for k in a.keys():
    print(k)
```

Value 리스트 만들기 - values

```python
a.values()
```

Key, Value 쌍 얻기 - items

items 함수는 Key와 Value의 쌍을 튜플로 묶은 값을 dict_items 객체로 반환한다.

```python
a.items()
```

Key: Value 쌍 모두 지우기 - clear

빈 리스트를 [], 빈 튜플을 ()로 표현하는 것과 마찬가지로 빈 딕셔너리도 {}로 표현한다.

```python
a.clear()
a
```

Key로 Value 얻기 - get

```python
a = {'name': 'pey', 'phone': '010-9999-1234', 'birth': '1118'}
a.get('name')
```

```python
a.get('phone')
```

딕셔너리에 존재하지 않는 키로 값을 가져오려고 할 경우, a['nokey'] 방식은 오류를 발생

```python
a = {'name':'pey', 'phone':'010-9999-1234', 'birth': '1118'}
print(a.get('nokey'))
```

```python
 a['nokey']
```

딕셔너리 안에 찾으려는 Key가 없을 경우, 미리 정해 둔 디폴트 값을 대신 가져오게 하고 싶을 때는 get(x, '디폴트 값')을 사용하면 편리하다.

```python
a.get('nokey', '정보없음')
```

해당 Key가 딕셔너리 안에 있는지 조사하기 - in

```python
a = {'name':'pey', 'phone':'010-9999-1234', 'birth': '1118'}
'name' in a
```

```python
'email' in a
```

Key로 Value 꺼내기 - pop

pop(x) 함수는 딕셔너리에서 Key가 x인 항목을 삭제한 후 그 Value를 반환한다.
get 함수와 달리 pop은 해당 항목을 딕셔너리에서 삭제한다는 차이가 있다.

```python
a = {'name':'pey', 'phone':'010-9999-1234', 'birth': '1118'}
phone = a.pop('phone')
phone
```
