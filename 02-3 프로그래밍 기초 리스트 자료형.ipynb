## 리스트(list) 자료형

* 파이썬에서는 리스트 자료형을 만들 때 `대괄호[]`로 감싸주고 그 안에 요소들을 `콤마(,)`로 구분해서 입력
* 파이썬에서는 리스트자료형은 String(str, 문자열)만큼 자주 사용되는 자료형임
> `변수명 = [요소1, 요소2, .... 요소n]`

#### 리스트(list)란?
1. 순서가 있는 자료형으로서 수정이 가능한 `객체의 집합`이다.
2. 수정, 삭제, 추가가 가능하다
3. list는 `대괄호[]`로 선언하고, 내부요소는 콤마로 구분한다.
4. list의 `인덱스는 0부터 시작`된다.
5. index의 범위가 벗어나는 경우에는 에러가 발생된다.
6. list는 어떠한 자료형도 함께 선언할 수 있다. 즉, 숫자, 문자, 함수, 리스트, 튜플, 문자열 등의 모든 자료형을 요소로 정의할 수 있다.
7. list는 `마지막 요소뒤에 콤마(,)`가 있어도 에러는 발생되지 않는다. 보통 생략을 하지만 편의를 위해서 콤마(,)를 추가하기도 한다.

리스트 확장 - extend

extend(x)에서 x에는 리스트만 올 수 있으며 원래의 a 리스트에 x 리스트를 더하게 된다.

a.extend([4, 5])는 a += [4, 5]와 동일하다. --> a += [4, 5]는 a = a + [4, 5]와 동일한 표현식이다.

```python
# 슬라이싱
# 문자열과 마찬가지로 리스트에서도 슬라이싱 기법을 사용할 수 있다.
# 파이썬에서는 리스트는 문법이 문자열에서 했던 방법과 동일
# [start:end:step]의 형태 사용

l = [1,2,3,4,5]

print(l[:])      # 처음:끝 --> [:] --> 전체
print(l[:1:2])   # 처음: 끝: Step
print(l[::3])
print()
print(l[1:3])  # 리스트
s = '12345'
print(s[1:3])  # 문자열
print()
print(l[:2])
print(l[2:])
print(l[2::2])
```

## 파이썬 리스트 & 문자열 슬라이싱 핵심 정리

파이썬의 슬라이싱(Slicing)은 리스트나 문자열에서 원하는 부분만 연속적으로 잘라내는 매우 강력하고 직관적인 문법입니다.

---

### 1. 슬라이싱 기본 문법

`[start : end : step]` 형태를 사용하며, 필요에 따라 각 요소는 생략할 수 있습니다.

* **start**: 추출을 시작할 인덱스 (포함, 생략 시 처음부터)
* **end**: 추출을 끝낼 인덱스 (**미포함**, 생략 시 끝까지)
* **step**: 인덱스의 증가 폭 (생략 시 기본값 1)

---

### 2. 코드 상세 분석 및 실행 결과

작성해주신 코드를 바탕으로 각 슬라이싱이 어떻게 동작하는지 정리했습니다.

**기준 데이터:**
* 리스트: `l = [1, 2, 3, 4, 5]`
* 문자열: `s = '12345'`

| 슬라이싱 코드 | 반환 결과 | 상세 설명 |
| :--- | :--- | :--- |
| `l[:]` | `[1, 2, 3, 4, 5]` | 처음부터 끝까지 전체를 추출합니다. |
| `l[:1:2]` | `[1]` | 처음부터 인덱스 1 미만까지 2칸씩 건너뛰며 추출합니다. (인덱스 0만 해당) |
| `l[::3]` | `[1, 4]` | 처음부터 끝까지 3칸씩 건너뛰며 추출합니다. (인덱스 0, 3) |
| `l[1:3]` | `[2, 3]` | 인덱스 1부터 인덱스 3 미만까지 추출합니다. |
| `s[1:3]` | `'23'` | **문자열**도 리스트와 완전히 동일한 원리로 동작합니다. |
| `l[:2]` | `[1, 2]` | 처음부터 인덱스 2 미만까지 추출합니다. |
| `l[2:]` | `[3, 4, 5]` | 인덱스 2부터 끝까지 추출합니다. |
| `l[2::2]` | `[3, 5]` | 인덱스 2부터 끝까지 2칸씩 건너뛰며 추출합니다. (인덱스 2, 4) |

---

```python
# 문제1) movie_rank 리스트에 '배트맨'을 추가하시오.
movie_rank = ['닥터 스트레인지', '스플릿', '럭키']
movie_rank.append('배트맨')
print(movie_rank)
```

```python
# 문제2) movie_rank 리스트의 '닥터 스트레인지'와 '스플릿'사이에
#'슈퍼맨'을 추가하시오
print(movie_rank)
movie_rank.insert(1,'슈퍼맨')
print(movie_rank)
```

```python
# 문제3) movie_rank 리스트에서 '럭키'를 삭제하시오.
print(movie_rank)
del movie_rank[3]
print(movie_rank)
```

```python
# 문제4) lang1과 lang2 리스트를 모두 갖는 langs 리스트를 만드시오
lang1 = ['C', 'C++', 'JAVA']
lang2 = ['Python', 'Go', 'C#']
langs = lang1 + lang2
print(langs)
```

```python
# 문제5) 리스트의 개수를 구하시오.

l = ['사과','배','포도','자두','복숭아']
print(len(l))
```

```python
# 문제 6) 리스트에서 최댓값, 최소값, 합계, 평균을 구하라.

l = [1, 2, 3, 5, 5, 10, 3, 7, 8, 6]

print('최대값 = ', max(l))
print('최소값 = ', min(l))
print('합계 = ', sum(l))
print('전체개수 = ', len(l))
print('평균 = ', sum(l) / len(l))     # 정수 / 정수 --> 실수
```
